```python
import os
import subprocess
from pathlib import Path

REPO_URL = 'https://github.com/<your-user>/<your-repo>.git'
REPO_DIR = Path('/content/news-stocks-python')

if REPO_DIR.exists():
    subprocess.run(['git', '-C', str(REPO_DIR), 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', REPO_URL, str(REPO_DIR)], check=True)

subprocess.run(['python', '-m', 'pip', 'install', '--upgrade', 'pip'], check=True)
subprocess.run(['python', '-m', 'pip', 'install', '-r', str(REPO_DIR / 'requirements.txt')], check=True)
print('Repo ready:', REPO_DIR)
```

```python
from pathlib import Path

drive_root = Path('/content/drive/MyDrive/news-stocks-python')
(drive_root / 'data/raw').mkdir(parents=True, exist_ok=True)
(drive_root / 'data/processed').mkdir(parents=True, exist_ok=True)
(drive_root / 'outputs/runs').mkdir(parents=True, exist_ok=True)
(drive_root / 'reports/run_manifests').mkdir(parents=True, exist_ok=True)

events_path = drive_root / 'data/raw/events.csv'
sample_events = Path('/content/news-stocks-python/data/raw/events.sample.csv')
if not events_path.exists() and sample_events.exists():
    events_path.write_text(sample_events.read_text(encoding='utf-8'), encoding='utf-8')

print('Drive data root:', drive_root)
print('Events CSV:', events_path)
```
